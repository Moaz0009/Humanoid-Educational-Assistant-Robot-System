```python
import sys
import time
import cv2
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QVBoxLayout
from PyQt5.QtCore import Qt, QTimer
from PyQt5.QtGui import QPixmap, QImage

# Assumed external modules
# from your_modules import detect_word, detect_verify, count_fingers_in_frame, whisper_asr, qwen_ar, text_to_speech_file, play_audio_file, serial_comm, detect_emotions_in_frame

class AIGUI(QWidget):
    def __init__(self, database):
        super().__init__()
        self.setWindowTitle("AI Teaching Assistant")
        self.showFullScreen()

        # Main label for UI
        self.label = QLabel("😴")
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        self.label.setAlignment(Qt.AlignCenter)

        layout = QVBoxLayout()
        layout.addWidget(self.label)
        self.setLayout(layout)

        # Core variables
        self.database = database
        self.identity = None
        self.cap = None
        self.video_timer = QTimer()

        # Q&A
        self.current_question = 0
        self.score = 0
        self.questions, self.answers = self._init_questions()

        # Finger detection
        self.required_consistency = 20
        self.last_finger_number = None
        self.finger_frame_count = 0
        self.true_finger_number = None

        # Start standby mode after launch
        QTimer.singleShot(1000, self.start_standby)

    def _init_questions(self):
        questions = [
            "ما هي عاصمة مصر؟",         # 0
            "ما هي عاصمة فرنسا؟",       # 1
            "أظهر الرقم 5",             # 2 ✅ Special
            "كم عدد أيام الأسبوع؟",     # 3
            "ما هو لون السماء؟",         # 4
            "أظهر الرقم 2",             # 5 ✅ Special
            "ما هو الحيوان الذي يقول موو؟", # 6
            "ما هو ناتج 3 + 2؟",         # 7
            "أظهر الرقم 7",             # 8 ✅ Special
            "ما هو اسم أطول نهر في العالم؟"  # 9
        ]
        answers = [
            "القاهرة",     # 0
            "باريس",       # 1
            "٥",           # 2
            "سبعة",        # 3
            "أزرق",        # 4
            "٢",           # 5
            "بقرة",        # 6
            "خمسة",        # 7
            "٧",           # 8
            "النيل"        # 9
        ]
        return questions, answers
    def start_standby(self):
        self.label.setText("😴")
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        QTimer.singleShot(1000, self.check_trigger_word)

    def check_trigger_word(self):
        if detect_word():
            self.label.setText("Trigger word detected. Verifying identity...")
            QTimer.singleShot(1000, self.verify_face)
        else:
            QTimer.singleShot(1000, self.check_trigger_word)

    def verify_face(self):
        self.label.setText("Initializing camera...")
        self.identity = "Unknown"
        self.start_time = time.time()
    
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
    
        self.video_timer.timeout.connect(self.update_camera_feed)
        self.video_timer.start(30)
    
    def update_camera_feed(self):
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)
    
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error.")
            self.video_timer.stop()
            return
    
        # ✅ Detect identity and get the annotated frame
        identity, annotated_frame = detect.detect_verify(frame, self.database)
    
        # ✅ Display the annotated frame
        rgb = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb.shape
        qt_img = QImage(rgb.data, w, h, ch * w, QImage.Format_RGB888)
        self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
        QApplication.processEvents()
    
        # ✅ If identity is known → move on
        elapsed = time.time() - self.start_time
        if identity != "Unknown":
            self.identity = identity
            self.video_timer.stop()
            self.cap.release()
            self.cap = None
            self.label.setText(f"Welcome {identity}!\nسوف  نبدا الاساله...")
            
            QTimer.singleShot(1500, self.ask_question)
    
        elif elapsed >= 5:
            self.video_timer.stop()
            self.cap.release()
            self.cap = None
            self.label.setText("No recognized face.\nReturning to standby.")
            QTimer.singleShot(3000, self.start_standby)
    def ask_question(self):
        # All questions completed
        if self.current_question >= len(self.questions):
            self.label.setText(f"Session complete!\n{self.identity}'s Score: {self.score}/{len(self.questions)}")
            self.current_question = 0
            self.score = 0
            QTimer.singleShot(5000, self.start_standby)
            return
    
        # Ask next question
        question = self.questions[self.current_question]
        self.label.setText(f"Q{self.current_question + 1}: {question}\n(Answer Now)")
        QApplication.processEvents()
    
        # 🔊 Play TTS audio
        audio_path = text_to_speech_file(text=question, output_path="question.mp3", lang='ar')
        play_audio_file(audio_path)
    
        # Special finger-based questions → detect fingers instead of speech
        if self.current_question in [2, 5, 8]:
            # Send expected number to Arduino (optional)
            expected_number = self.answers[self.current_question]
            serial_comm.send(expected_number)
    
            # Start finger count session
            self.finger_frame_count = 0
            self.last_finger_number = None
            self.true_finger_number = None
    
            self.start_finger_check()
        else:
            # Start speech answer recording after a short delay
            QTimer.singleShot(1000, self.record_answer)
    def start_finger_check(self):
        self.finger_timer = QTimer()
        self.finger_timer.timeout.connect(self._check_fingers)
        self.finger_timer.start(100)  # Every 100ms (~10fps)
    def _check_fingers(self):
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error during finger check.")
            self.finger_timer.stop()
            return
    
        # Run finger detection
        frame, current_number = count_fingers_in_frame(frame)
    
        # Stability check
        if current_number == self.last_finger_number:
            self.finger_frame_count += 1
        else:
            self.finger_frame_count = 1
            self.last_finger_number = current_number
    
        # Show updated frame with fingers
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb.shape
        qt_img = QImage(rgb.data, w, h, ch * w, QImage.Format_RGB888)
        self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
    
        # Confirmed after 20 consistent frames
        if self.finger_frame_count >= self.required_consistency:
            self.finger_timer.stop()
            self.true_finger_number = current_number
    
            correct_answer = int(self.answers[self.current_question])
            if self.true_finger_number == correct_answer:
                self.label.setText("✅ Correct!")
                play_audio_file(correct_path)
                self.score += 1
            else:
                self.label.setText("❌ Incorrect.")
                play_audio_file(false_number)
    
            self.current_question += 1
            QTimer.singleShot(3000, self.ask_question)
    
    def record_answer(self):
        self.label.setText("🎤 Listening...")
        QApplication.processEvents()
    
        # Record and transcribe
        answer = whisper_asr.transcribe_audio()
        self.answer_to_check = answer.strip()
    
        if not self.answer_to_check:
            self.label.setText("Didn't catch that. Please try again.")
            QTimer.singleShot(2000, self.ask_question)
            return
    
        self.label.setText("Processing...")
        QApplication.processEvents()
    
        # Proceed to verification and emotion analysis
        self.start_identity_verification(attempts=5, rounds=3)
    def start_identity_verification(self, attempts=5, rounds=3):
        self.verification_attempts = attempts
        self.verification_rounds = rounds
        self.successful_identity = False
        self._verify_identity()
    def _verify_identity(self):
        if self.verification_attempts == 0:
            self.verification_rounds -= 1
            if self.successful_identity:
                self._start_check_emotion()
                return
            elif self.verification_rounds == 0:
                self.label.setText("Identity mismatch. Returning to standby.")
                QTimer.singleShot(2000, self.start_standby)
                return
            else:
                self.verification_attempts = 5
                QTimer.singleShot(500, self._verify_identity)
                return
    
        # Read frame and verify again
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)
    
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error during identity check.")
            QTimer.singleShot(2000, self.start_standby)
            return
    
        identity_check, _ = detect.detect_verify(frame, self.database)
        if identity_check == self.identity:
            self.successful_identity = True
            self._start_check_emotion()
            return
    
        self.verification_attempts -= 1
        QTimer.singleShot(500, self._verify_identity)
    def _start_check_emotion(self):
        self.emotion_attempts = 5
        self.detected_emotion = "Unknown"
        self._check_emotion()
    def _check_emotion(self):
        if self.emotion_attempts == 0:
            self._finalize_answer_with_emotion()
            return
    
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)
    
        ret, frame = self.cap.read()
        if ret:
            try:
                _, emotions = detect_emotions_in_frame(frame)
                if emotions and emotions[0].lower() != "unknown":
                    self.detected_emotion = emotions[0]
                    self._finalize_answer_with_emotion()
                    return
            except Exception as e:
                print(f"Emotion detection error: {e}")
    
        self.emotion_attempts -= 1
        QTimer.singleShot(500, self._check_emotion)
    def _finalize_answer_with_emotion(self):
        emotion = self.detected_emotion.lower()
        if emotion in ["sad", "angry", "fear"]:
            self.label.setText(f"Detected emotion: {emotion}. Calling an adult...")
            self.get_adult()
            return
    
        # ✅ Otherwise: Evaluate answer
        correct = qwen_ar.evaluate_answer(self.answers[self.current_question], self.answer_to_check)
        if correct:
            self.label.setText("✅ Correct!")
            play_audio_file(correct_path)
            self.score += 1
        else:
            self.label.setText("❌ Incorrect.")
            play_audio_file(false_ques)
    
        self.current_question += 1
        QTimer.singleShot(3000, self.ask_question)
    def get_adult(self):
        self.adult_target = "Mr.Ali"
        self.adult_check_timer = QTimer()
        self.adult_check_timer.timeout.connect(self._check_for_adult)
        self.adult_check_timer.start(1000)  # every 1 second
    
    def _check_for_adult(self):
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)
    
        play_audio_file(adult)  # play alert
    
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error during adult check.")
            self.adult_check_timer.stop()
            return
    
        name, annotated = detect.detect_verify(frame, self.database)
    
        # Show result on screen
        rgb = cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb.shape
        qt_img = QImage(rgb.data, w, h, ch * w, QImage.Format_RGB888)
        self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
    
        # Found the adult
        if name == self.adult_target:
            self.adult_check_timer.stop()
            if self.cap and self.cap.isOpened():
                self.cap.release()
                self.cap = None
            self.label.setText("Adult detected. Returning to standby.")
            QTimer.singleShot(3000, self.start_standby)
import sys
import os
from PyQt5.QtWidgets import QApplication

# Optional: to avoid OpenGL crash issues
os.environ["QT_QUICK_BACKEND"] = "software"

# Only run this block if in the main context
if __name__ == '__main__':
    # Prevent multiple QApplication instances
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)

    # ✅ Load your face database here
    # Example: from your_modules import load_face_database
    # database = load_face_database()

    # ✅ Initialize and show the GUI
    gui = AIGUI(database)
    gui.show()

    # ✅ Use QTimer to delay app.exec_() safely in Jupyter
    def run_app():
        app.exec_()

    from PyQt5.QtCore import QTimer
    QTimer.singleShot(0, run_app)

```

```python
import sys
import cv2
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QVBoxLayout
from PyQt5.QtCore import Qt, QTimer
from PyQt5.QtGui import QImage, QPixmap

class LiveCamera(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Live Camera Feed")
        self.setStyleSheet("background-color: black;")
        self.setGeometry(100, 100, 800, 600)

        self.label = QLabel(self)
        self.label.setAlignment(Qt.AlignCenter)

        layout = QVBoxLayout()
        layout.addWidget(self.label)
        self.setLayout(layout)

        self.cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)  # CAP_DSHOW fixes many Windows bugs

        if not self.cap.isOpened():
            self.label.setText("Camera not found.")
            return

        self.timer = QTimer()
        self.timer.timeout.connect(self.update_frame)
        self.timer.start(30)

    def update_frame(self):
        ret, frame = self.cap.read()
        if not ret or frame is None:
            self.label.setText("Failed to read frame.")
            return

        # Convert to RGB
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb.shape
        bytes_per_line = ch * w
        qt_img = QImage(rgb.data, w, h, bytes_per_line, QImage.Format_RGB888)
        pixmap = QPixmap.fromImage(qt_img)
        self.label.setPixmap(pixmap.scaled(self.label.size(), Qt.KeepAspectRatio))

    def closeEvent(self, event):
        self.cap.release()
        event.accept()

# Launch safely
app = QApplication.instance()
if app is None:
    app = QApplication(sys.argv)

camera_window = LiveCamera()
camera_window.show()
```

```python
import sys
import time
import cv2
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QVBoxLayout
from PyQt5.QtCore import QTimer
from PyQt5.QtGui import QPixmap, QImage

# Placeholder imports for your system modules
#from your_modules import detect_word, detect_verify, whisper_asr, qwen_evaluate  # Replace with actual module paths

class AIGUI(QWidget):
 from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QVBoxLayout
from PyQt5.QtCore import Qt, QTimer

class AIGUI(QWidget):
    def __init__(self):
        super().__init__()

        self.setWindowTitle("AI Teaching Assistant")
        self.showFullScreen()

        # ✅ Main label
        self.label = QLabel("😴", self)
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        self.label.setAlignment(Qt.AlignCenter)
        self.label.setWordWrap(True)

        layout = QVBoxLayout()
        layout.addWidget(self.label)
        self.setLayout(layout)

        # ✅ Question/answer logic
        self.questions = [
            "ما هي عاصمة مصر؟",         # 0
            "ما هي عاصمة فرنسا؟",       # 1
            "أظهر الرقم 5",             # 2 ✅ Special
            "كم عدد أيام الأسبوع؟",     # 3
            "ما هو لون السماء؟",         # 4
            "أظهر الرقم 2",             # 5 ✅ Special
            "ما هو الحيوان الذي يقول موو؟", # 6
            "ما هو ناتج 3 + 2؟",         # 7
            "أظهر الرقم 7",             # 8 ✅ Special
            "ما هو اسم أطول نهر في العالم؟"  # 9
        ]
        
        self.answers = [
            "القاهرة",     # 0
            "باريس",       # 1
            "٥",           # 2 ✅
            "سبعة",        # 3
            "أزرق",        # 4
            "٢",           # 5 ✅
            "بقرة",        # 6
            "خمسة",        # 7
            "٧",           # 8 ✅
            "النيل"        # 9
        ]

        self.current_question = 0
        self.identity = None
        self.score = 0
        self.last_finger_number = None
        self.finger_frame_count = 0
        self.true_finger_number = None
        self.required_consistency = 20  # number of frames to confirm

        self.database = database  # your actual face DB
        self.video_timer = QTimer()
        self.cap = None

        # ✅ Start standby after UI loads
        QTimer.singleShot(1000, self.start_standby)

    def start_standby(self):
        self.label.setText("😴")
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        QTimer.singleShot(1000, self.check_trigger_word)


    def check_trigger_word(self):
        if detect_word():
            self.label.setText("Trigger word detected. Verifying identity...")
            self.verify_face()
        else:
            QTimer.singleShot(1000, self.check_trigger_word)

    def verify_face(self):
        self.label.setText("Initializing camera...")
        self.identity = "Unknown"
        self.cap = cv2.VideoCapture(0)
        self.start_time = time.time()
        self.video_timer.timeout.connect(self.update_camera_feed)
        self.video_timer.start(30)
        self.cap.release()


    def update_camera_feed(self):
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)

            
        
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error.1")
            self.video_timer.stop()
            return
    
        # Show raw feed regardless of detection
        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb_frame.shape
        bytes_per_line = ch * w
        qt_img = QImage(rgb_frame.data, w, h, bytes_per_line, QImage.Format_RGB888)
        self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
    
        # Only start verifying identity after 5 seconds
        elapsed = time.time() - self.start_time
    
        if elapsed >= 5 and self.identity == "Unknown":
            identity, processed_frame = detect.detect_verify(frame, self.database)
            self.identity = identity
    
            if identity != "Unknown":
                self.video_timer.stop()
                self.cap.release()
                self.label.setText(f"Welcome {identity}!\nStarting Q&A")
                QTimer.singleShot(1000, self.ask_question)
            elif elapsed >= 10:  # total timeout (5 seconds view + 5 seconds verification)
                self.video_timer.stop()
                self.cap.release()
                self.label.setText("No recognized face.\nReturning to standby.")
                QTimer.singleShot(2000, self.start_standby)
    def Finger_Count(self):
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error.2")
            self.video_timer.stop()
            return
    
        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb_frame.shape
        bytes_per_line = ch * w
    
        rgb_frame, current_finger_number = count_fingers_in_frame(frame)
    
        # Check if the number is same as last frame
        if current_finger_number == self.last_finger_number:
            self.finger_frame_count += 1
        else:
            self.finger_frame_count = 1  # reset counter
            self.last_finger_number = current_finger_number
    
        # When stable for enough frames → accept as true result
        if self.finger_frame_count >= self.required_consistency:
            self.true_finger_number = current_finger_number
            # you can now display or use true_finger_number
            print(f"Confirmed finger number: {self.true_finger_number}")
    
        # Update video feed
        qt_img = QImage(rgb_frame.data, w, h, bytes_per_line, QImage.Format_RGB888)
        self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))

    def ask_question(self):
        if self.current_question >= len(self.questions):
            self.label.setText(f"Session complete!\n{self.identity}'s Score: {self.score}/{len(self.questions)}")
            self.current_question = 0
            self.score = 0
            QTimer.singleShot(5000, self.start_standby)
            return
    
        question = self.questions[self.current_question]
        self.label.setText(f"Q{self.current_question+1}: {question}\n(Answer Now)")
        QApplication.processEvents()  # ✅ force UI to update immediately
        print(question)
        # ✅ Play question using TTS
    
        question = self.questions[self.current_question]
        self.label.setText(f"Q{self.current_question+1}: {question}\n(Answer Now)")
        QApplication.processEvents()  # ✅ force UI to update immediately
        print(question)
        # ✅ Play question using TTS
        audio_path=text_to_speech_file(text=question,
                           output_path="question.mp3",
                           lang='ar')
        play_audio_file('question.mp3') ################################################
        print('workiong')
        # ✅ Special treatment for index 2, 5, 8 (0-based)
        if self.current_question in [2, 5, 8]:
            serial_comm.send(question[-1])
            fingers=self.Finger_Count()##############################
            if (fingers==self.current_question[-1]):
                self.label.setText("Correct!")
                play_audio_file(correct_path)
                self.score += 1
            else:
                self.label.setText("Sorry, false answer.")
                play_audio_file(false_number)
            self.current_question += 1
            QTimer.singleShot(3000, self.ask_question)
                
        else:
            # ✅ Move to answer recording after short delay
            QTimer.singleShot(1000, self.record_answer)

    def record_answer(self):
        answer = whisper_asr.transcribe_audio()
        if not answer.strip():
            self.label.setText("Didn't catch that, please repeat")
        # ✅ Special treatment for index 2, 5, 8 (0-based)
        if self.current_question in [2, 5, 8]:
            serial_comm.send(question[-1])
            fingers=self.Finger_Count()##############################
            if (fingers==self.current_question[-1]):
                self.label.setText("Correct!")
                play_audio_file(correct_path)
                self.score += 1
            else:
                self.label.setText("Sorry, false answer.")
                play_audio_file(false_number)
            self.current_question += 1
            QTimer.singleShot(3000, self.ask_question)
                
        else:
            # ✅ Move to answer recording after short delay
            QTimer.singleShot(1000, self.record_answer)
            

  ## def record_answer(self):
    ##    answer = whisper_asr.transcribe_audio()
      ##  if not answer.strip():
        ##    self.label.setText("Didn't catch that. Please try again.")
           ## QTimer.singleShot(2000, self.ask_question)
            ##return
    
        ##self.label.setText("Evaluating...")
        ##QApplication.processEvents()  # ✅ Ensure GUI updates immediately
    
        ##self.answer_to_check = answer  # ✅ Save the answer to use later
      ##  self.start_identity_verification(attempts=5, rounds=3)  # ✅ Begin identity check
    def start_identity_verification(self, attempts=5, rounds=3):
        self.verification_attempts = attempts
        self.verification_rounds = rounds
        self.successful_identity = False
        self._verify_identity()
    
    def _verify_identity(self):
        if self.verification_attempts == 0:
            self.verification_rounds -= 1
            if self.successful_identity:
                self._finish_answer_check()
                return
            elif self.verification_rounds == 0:
                self.label.setText("Identity mismatch. Returning to standby.")
                QTimer.singleShot(2000, self.start_standby)
                return
            else:
                self.verification_attempts = 5  # Reset for next round
                self.successful_identity = False
                QTimer.singleShot(500, self._verify_identity)
                return
                
            #if self.cap is None or not self.cap.isOpened():
                #self.cap = cv2.VideoCapture(0)
                #time.sleep(0.5)
        
                ret, frame_check = self.cap.read()
    
                if not ret:
                    self.label.setText("Camera error during identity check.")
                    QTimer.singleShot(2000, self.start_standby)
                    return
        
                identity_check, _ = detect.detect_verify(frame_check, self.database)
            
                if identity_check == self.identity:
                    self.successful_identity = True
                    self._start_check_emotion()  # ✅ Early exit when match is found
                    return
            
            self.verification_attempts -= 1
            QTimer.singleShot(500, self._verify_identity)

    def _start_check_emotion(self):
        self.emotion_attempts = 5
        self.detected_emotion = "Unknown"
        self._check_emotion()

        
    def get_adult(self):
        
        adult_name=""   
        if self.cap is None or not self.cap.isOpened():
            
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)
            
        while(adult_name!="Mr.Ali"):
            play_audio_file(adult)
            
            ret, frame = self.cap.read()
            
            if not ret:
                self.label.setText("Camera error3.")
                self.video_timer.stop()
                return
        
            # Show raw feed regardless of detection
            rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            h, w, ch = rgb_frame.shape
            bytes_per_line = ch * w
            adult_name, rgb_frame = detect.detect_verify(frame, self.database)
            qt_img = QImage(rgb_frame.data, w, h, bytes_per_line, QImage.Format_RGB888)
            self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
        if self.cap and self.cap.isOpened():
            self.cap.release()
            self.cap = None
            
        QTimer.singleShot(5000, self.start_standby)
        

        
    def _check_emotion(self):
        if self.emotion_attempts == 0:
            # Emotion not found after retries
            self._finalize_answer_with_emotion()
            return
        if self.cap is None or not self.cap.isOpened():####
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)  # Optional delay for camera warm-up
    
        ret, frame = self.cap.read()
        if ret:
            try:
              _, emotion = detect_emotions_in_frame(frame)  # emotion is a list
              if emotion and len(emotion) > 0:
                 first_emotion = emotion[0]
                 if first_emotion and first_emotion.lower() != "unknown":
                    self.detected_emotion = first_emotion
                    self._finalize_answer_with_emotion()  # ✅ Early exit if emotion is valid
                    return
            except Exception as e:
                print(f"Emotion detection error: {e}")
                  
        self.emotion_attempts -= 1
        QTimer.singleShot(500, self._check_emotion)  # Retry after delay

    def _finalize_answer_with_emotion(self):
        emotion = self.detected_emotion.lower()
    
        # 🚨 React to emotional state
        if emotion in ["sad", "angry", "fear"]:
            self.label.setText(f"Detected emotion: {emotion}. Getting an adult...")
            self.get_adult()##May be we play an Alarming Sound so that an adult can come and deal with it.
            QTimer.singleShot(5000, self.start_standby)  # Exit after response
            return
    
        # ✅ Normal answer handling
        correct = qwen_ar.evaluate_answer(self.answers[self.current_question], self.answer_to_check)
        if correct:
            self.label.setText(f"Correct!")
            play_audio_file(correct_path)
            self.score += 1
        else:
            self.label.setText(f"Sorry, false answer.")
            play_audio_file(false_ques)
    
        self.current_question += 1
        QTimer.singleShot(3000, self.ask_question)

if __name__ == '__main__':
    app = QApplication(sys.argv)
    window = AIGUI()
    window.show()
    sys.exit(app.exec_())
```

```python

import sys
import time
import cv2
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QVBoxLayout
from PyQt5.QtCore import QTimer, Qt
from PyQt5.QtGui import QPixmap, QImage
import threading

class AIGUI(QWidget):
    from PyQt5.QtCore import Qt, QTimer

    def __init__(self):
        super().__init__()

        self.setWindowTitle("AI Teaching Assistant")
        self.showFullScreen()

        self.label = QLabel("😴", self)
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        self.label.setAlignment(Qt.AlignCenter)
        self.label.setWordWrap(True)

        layout = QVBoxLayout()
        layout.addWidget(self.label)
        self.setLayout(layout)

        self.questions = [
            "ما هي عاصمة مصر؟",
            "ما هي عاصمة فرنسا؟",
            "أظهر الرقم 5",
            "كم عدد أيام الأسبوع؟",
            "ما هو لون السماء؟",
            "أظهر الرقم 2",
            "ما هو الحيوان الذي يقول موو؟",
            "ما هو ناتج 3 + 2؟",
            "أظهر الرقم 7",
            "ما هو اسم أطول نهر في العالم؟"
        ]

        self.answers = [
            "القاهرة",
            "باريس",
            "٥",
            "سبعة",
            "أزرق",
            "٢",
            "بقرة",
            "خمسة",
            "٧",
            "النيل"
        ]

        self.current_question = 0
        self.identity = None
        self.score = 0
        self.last_finger_number = None
        self.finger_frame_count = 0
        self.true_finger_number = None
        self.required_consistency = 20

        self.database = database
        self.video_timer = QTimer()
        self.cap = None

        QTimer.singleShot(1000, self.start_standby)

    def start_standby(self):
        if self.cap and self.cap.isOpened():
            self.cap.release()
            self.cap = None
        self.label.setText("😴")
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        QTimer.singleShot(1000, self.check_trigger_word)

    def check_trigger_word(self):
        if detect_word():
            self.label.setText("Trigger word detected. Verifying identity...")
            self.verify_face()
        else:
            QTimer.singleShot(1000, self.check_trigger_word)

    def verify_face(self):
        self.label.setText("Initializing camera...")
        self.identity = "Unknown"
        self.start_time = time.time()
    
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
    
        self.video_timer.timeout.connect(self.update_camera_feed)
        self.video_timer.start(30)
    
    def update_camera_feed(self):
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)
    
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error.")
            self.video_timer.stop()
            return
    
        # ✅ Detect identity and get the annotated frame
        identity, annotated_frame = detect.detect_verify(frame, self.database)
    
        # ✅ Display the annotated frame
        rgb = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb.shape
        qt_img = QImage(rgb.data, w, h, ch * w, QImage.Format_RGB888)
        self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
        QApplication.processEvents()
    
        # ✅ Always wait full 5 seconds before deciding
        elapsed = time.time() - self.start_time
        if elapsed >= 5:
            self.video_timer.stop()
            self.cap.release()
            self.cap = None
    
            if identity != "Unknown":
                self.identity = identity
                self.label.setText(f"Welcome {identity}!\nStarting Q&A...")
                play_audio_file('hello_message.mp3')
                QTimer.singleShot(1500, self.ask_question)
            else:
                self.label.setText("ﻻ يوجد طﻻب هنا.\nالعوده للانتظار.")
                QTimer.singleShot(3000, self.start_standby)
    
    def Finger_Count(self):
        print('In Finger_Count...')
        self.finger_frame_count = 0
        self.last_finger_number = None
        start_time = time.time()
    
        while time.time() - start_time < 7:  # max 5 seconds
            ret, frame = self.cap.read()
            if not ret:
                self.label.setText("Camera error during finger count.")
                return None
    
            rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            h, w, ch = rgb_frame.shape
            bytes_per_line = ch * w
    
            rgb_frame, current_finger_number = count_fingers_in_frame(rgb_frame)
    
            qt_img = QImage(rgb_frame.data, w, h, bytes_per_line, QImage.Format_RGB888)
            self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
            QApplication.processEvents()
    
            if current_finger_number == self.last_finger_number:
                self.finger_frame_count += 1
            else:
                self.finger_frame_count = 1
                self.last_finger_number = current_finger_number
    
            if self.finger_frame_count >= self.required_consistency:
                print(f"Confirmed finger number: {current_finger_number}")
                return current_finger_number
    
            time.sleep(0.05)  # Avoid CPU overload
    
        print("Finger count timeout")
        return None
    def ask_question(self):
        if self.current_question >= len(self.questions):
            self.label.setText(f"لقد انتهت جلستنا!\n{self.identity} الدرجه: {self.score}/{len(self.questions)}")
            play_audio_file('goodbye_message.mp3')
            self.current_question = 0
            self.score = 0
            QTimer.singleShot(5000, self.start_standby)
            return

        question = self.questions[self.current_question]
        self.label.setText(f"Q{self.current_question+1}: {question}\n(Answer Now)")
        QApplication.processEvents()

        audio_path = text_to_speech_file(text=question, output_path="question.mp3", lang='ar')
        play_audio_file('question.mp3')

        if self.current_question in [2, 5, 8]:
            print('number')
            serial_comm.send(question[-1])
            fingers = self.Finger_Count()  # ✅ USE return value
        
            try:
                expected_digit = int(question[-1])
            except:
                expected_digit = -1
        
            if fingers is not None and fingers == expected_digit:
                self.label.setText("اجابه صحيحه!")
                play_audio_file(correct_path)
                self.score += 1
            else:
                self.label.setText("اجابه خاطئه برجاء التركيز.")
                play_audio_file(false_number)
        
            self.current_question += 1
            QTimer.singleShot(3000, self.ask_question)
        else:
            QTimer.singleShot(1000, self.record_answer)
    def record_answer(self):
        answer = whisper_asr.transcribe_audio()
        if  answer.strip()== 'ترجمة نانسي قنقر':
            self.label.setText("لم اسمع ذلك بوضوح, برجاء الاعاده")
            QTimer.singleShot(2000, self.ask_question)
            return

        self.label.setText("جاري التصحيح...")
        QApplication.processEvents()

        self.answer_to_check = answer
        self.start_identity_verification(attempts=5, rounds=3)
    def start_identity_verification(self, attempts=5, rounds=3):
        self.verification_attempts = attempts
        self.verification_rounds = rounds
        self.successful_identity = False
        self._verify_identity()

    def _verify_identity(self):
        if self.verification_attempts == 0:
            self.verification_rounds -= 1
            if self.successful_identity:
                self._finish_answer_check()
                return
            elif self.verification_rounds == 0:
                self.label.setText("لست نفس الشخص , انهاء الجلسه")
                QTimer.singleShot(2000, self.start_standby)
                return
            else:
                self.verification_attempts = 1
                self.successful_identity = False
                QTimer.singleShot(500, self._verify_identity)
                return
    
        # ✅ Initialize the camera if needed
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
            time.sleep(0.5)
    
        ret, frame_check = self.cap.read()
    
        if not ret:
            self.label.setText("Camera error during identity check.")
            QTimer.singleShot(2000, self.start_standby)
            return
    
        identity_check, _ = detect.detect_verify(frame_check, self.database)
    
        if identity_check == self.identity:
            self.successful_identity = True
            self._start_check_emotion()
            return
    
        self.verification_attempts -= 1
        QTimer.singleShot(500, self._verify_identity)
    def _start_check_emotion(self):
        self.emotion_attempts = 1
        self.detected_emotion = "Unknown"
        self._check_emotion()

    def _check_emotion(self):
        if self.emotion_attempts == 0:
            self._finalize_answer_with_emotion()
            return

        ret, frame = self.cap.read()
        if ret:
            try:
                _, emotion = detect_emotions_in_frame(frame)
                if emotion and len(emotion) > 0:
                    first_emotion = emotion[0]
                    if first_emotion and first_emotion.lower() != "unknown":
                        self.detected_emotion = first_emotion
                        self._finalize_answer_with_emotion()
                        return
            except Exception as e:
                print(f"Emotion detection error: {e}")

        self.emotion_attempts -= 1
        QTimer.singleShot(500, self._check_emotion)


    def get_adult(self):
        adult_name = ""
    
        # Start the camera if it was released
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(0)
    
        def play_adult_audio():
            play_audio_file(adult)
    
        while adult_name != "Mr.Ali":
            threading.Thread(target=play_adult_audio).start()
    
            ret, frame = self.cap.read()
            if not ret:
                self.label.setText("Camera error3.")
                self.video_timer.stop()
                return
    
            adult_name, annotated_frame = detect.detect_verify(frame, self.database)
            rgb_frame = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)
            h, w, ch = rgb_frame.shape
            bytes_per_line = ch * w
            qt_img = QImage(rgb_frame.data, w, h, bytes_per_line, QImage.Format_RGB888)
            self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
            QApplication.processEvents()
            time.sleep(0.5)  # slight delay to prevent excessive frame grabbing
    
        self.cap.release()
        self.cap = None
        QTimer.singleShot(5000, self.start_standby)
    def _finalize_answer_with_emotion(self):
        emotion = self.detected_emotion.lower()

        if emotion in ["angry", "fear"]:
            self.label.setText(f"Detected emotion: {emotion}. Getting an adult...")
            self.get_adult()
            QTimer.singleShot(5000, self.start_standby)
            return

        correct = qwen_ar.evaluate_answer(self.answers[self.current_question], self.answer_to_check)
        if correct =='1':
            self.label.setText("اجابه صحيحيه!")
            play_audio_file(correct_path)
            self.score += 1
        else:
            self.label.setText("اجابه خاطئه , برجاء التركيز.")
            play_audio_file(false_ques)

        self.current_question += 1
        QTimer.singleShot(3000, self.ask_question)
if __name__ == '__main__':
    app = QApplication(sys.argv)
    window = AIGUI()
    window.show()
    sys.exit(app.exec_())
```

```python

import sys
import time
import cv2
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QVBoxLayout
from PyQt5.QtCore import QTimer, Qt
from PyQt5.QtGui import QPixmap, QImage
import threading
def gstreamer_pipeline(
    capture_width=1280,
    capture_height=720,
    display_width=1280,
    display_height=720,
    framerate=30,
    flip_method=0,
):
    return (
        "nvarguscamerasrc ! "
        "video/x-raw(memory:NVMM), width=(int){}, height=(int){}, "
        "format=(string)NV12, framerate=(fraction){}/1 ! "
        "nvvidconv flip-method={} ! "
        "video/x-raw, width=(int){}, height=(int){}, format=(string)BGRx ! "
        "videoconvert ! "
        "video/x-raw, format=(string)BGR ! appsink"
        .format(
            capture_width,
            capture_height,
            framerate,
            flip_method,
            display_width,
            display_height,
        )
    )

class AIGUI(QWidget):
    from PyQt5.QtCore import Qt, QTimer

    def __init__(self):
        super().__init__()

        self.setWindowTitle("AI Teaching Assistant")
        self.showFullScreen()

        self.label = QLabel("😴", self)
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        self.label.setAlignment(Qt.AlignCenter)
        self.label.setWordWrap(True)

        layout = QVBoxLayout()
        layout.addWidget(self.label)
        self.setLayout(layout)

        self.questions = [
            "ما هي عاصمة مصر؟",
            "ما هي عاصمة فرنسا؟",
            "أظهر الرقم 5",
            "كم عدد أيام الأسبوع؟",
            "ما هو لون السماء؟",
            "أظهر الرقم 2",
            "ما هو الحيوان الذي يقول موو؟",
            "ما هو ناتج 3 + 2؟",
            "أظهر الرقم 7",
            "ما هو اسم أطول نهر في العالم؟"
        ]

        self.answers = [
            "القاهرة",
            "باريس",
            "٥",
            "سبعة",
            "أزرق",
            "٢",
            "بقرة",
            "خمسة",
            "٧",
            "النيل"
        ]

        self.current_question = 0
        self.identity = None
        self.score = 0
        self.last_finger_number = None
        self.finger_frame_count = 0
        self.true_finger_number = None
        self.required_consistency = 20

        self.database = database
        self.video_timer = QTimer()
        self.cap = None

        QTimer.singleShot(1000, self.start_standby)

    def start_standby(self):
        if self.cap and self.cap.isOpened():
            self.cap.release()
            self.cap = None
        self.label.setText("😴")
        self.label.setStyleSheet("font-size: 150px; color: white; background-color: black;")
        QTimer.singleShot(1000, self.check_trigger_word)

    def check_trigger_word(self):
        if detect_word():
            self.label.setText("Trigger word detected. Verifying identity...")
            self.verify_face()
        else:
            QTimer.singleShot(1000, self.check_trigger_word)

    def verify_face(self):
        self.label.setText("Initializing camera...")
        self.identity = "Unknown"
        self.start_time = time.time()
    
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(gstreamer_pipeline(), cv2.CAP_GSTREAMER)
    
        self.video_timer.timeout.connect(self.update_camera_feed)
        self.video_timer.start(30)
    
    def update_camera_feed(self):
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(gstreamer_pipeline(), cv2.CAP_GSTREAMER)
            time.sleep(0.5)
    
        ret, frame = self.cap.read()
        if not ret:
            self.label.setText("Camera error.")
            self.video_timer.stop()
            return
    
        # ✅ Detect identity and get the annotated frame
        identity, annotated_frame = detect.detect_verify(frame, self.database)
    
        # ✅ Display the annotated frame
        rgb = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb.shape
        qt_img = QImage(rgb.data, w, h, ch * w, QImage.Format_RGB888)
        self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
        QApplication.processEvents()
    
        # ✅ Always wait full 5 seconds before deciding
        elapsed = time.time() - self.start_time
        if elapsed >= 5:
            self.video_timer.stop()
            self.cap.release()
            self.cap = None
    
            if identity != "Unknown":
                self.identity = identity
                self.label.setText(f"Welcome {identity}!\nStarting Q&A...")
                play_audio_file('hello_message.mp3')
                QTimer.singleShot(1500, self.ask_question)
            else:
                self.label.setText("ﻻ يوجد طﻻب هنا.\nالعوده للانتظار.")
                QTimer.singleShot(3000, self.start_standby)
    
    def Finger_Count(self):
        print('In Finger_Count...')
        self.finger_frame_count = 0
        self.last_finger_number = None
        start_time = time.time()
    
        while time.time() - start_time < 7:  # max 5 seconds
            ret, frame = self.cap.read()
            if not ret:
                self.label.setText("Camera error during finger count.")
                return None
    
            rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            h, w, ch = rgb_frame.shape
            bytes_per_line = ch * w
    
            rgb_frame, current_finger_number = count_fingers_in_frame(rgb_frame)
    
            qt_img = QImage(rgb_frame.data, w, h, bytes_per_line, QImage.Format_RGB888)
            self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
            QApplication.processEvents()
    
            if current_finger_number == self.last_finger_number:
                self.finger_frame_count += 1
            else:
                self.finger_frame_count = 1
                self.last_finger_number = current_finger_number
    
            if self.finger_frame_count >= self.required_consistency:
                print(f"Confirmed finger number: {current_finger_number}")
                return current_finger_number
    
            time.sleep(0.05)  # Avoid CPU overload
    
        print("Finger count timeout")
        return None
    def ask_question(self):
        if self.current_question >= len(self.questions):
            self.label.setText(f"لقد انتهت جلستنا!\n{self.identity} الدرجه: {self.score}/{len(self.questions)}")
            play_audio_file('goodbye_message.mp3')
            self.current_question = 0
            self.score = 0
            QTimer.singleShot(5000, self.start_standby)
            return

        question = self.questions[self.current_question]
        self.label.setText(f"Q{self.current_question+1}: {question}\n(Answer Now)")
        QApplication.processEvents()

        audio_path = text_to_speech_file(text=question, output_path="question.mp3", lang='ar')
        play_audio_file('question.mp3')

        if self.current_question in [2, 5, 8]:
            print('number')
            serial_comm.send(question[-1])
            fingers = self.Finger_Count()  # ✅ USE return value
        
            try:
                expected_digit = int(question[-1])
            except:
                expected_digit = -1
        
            if fingers is not None and fingers == expected_digit:
                self.label.setText("اجابه صحيحه!")
                play_audio_file(correct_path)
                self.score += 1
            else:
                self.label.setText("اجابه خاطئه برجاء التركيز.")
                play_audio_file(false_number)
        
            self.current_question += 1
            QTimer.singleShot(3000, self.ask_question)
        else:
            QTimer.singleShot(1000, self.record_answer)
    def record_answer(self):
        answer = whisper_asr.transcribe_audio()
        if  answer.strip()== 'ترجمة نانسي قنقر':
            self.label.setText("لم اسمع ذلك بوضوح, برجاء الاعاده")
            QTimer.singleShot(2000, self.ask_question)
            return

        self.label.setText("جاري التصحيح...")
        QApplication.processEvents()

        self.answer_to_check = answer
        self.start_identity_verification(attempts=5, rounds=3)
    def start_identity_verification(self, attempts=5, rounds=3):
        self.verification_attempts = attempts
        self.verification_rounds = rounds
        self.successful_identity = False
        self._verify_identity()

    def _verify_identity(self):
        if self.verification_attempts == 0:
            self.verification_rounds -= 1
            if self.successful_identity:
                self._finish_answer_check()
                return
            elif self.verification_rounds == 0:
                self.label.setText("لست نفس الشخص , انهاء الجلسه")
                QTimer.singleShot(2000, self.start_standby)
                return
            else:
                self.verification_attempts = 1
                self.successful_identity = False
                QTimer.singleShot(500, self._verify_identity)
                return
    
        # ✅ Initialize the camera if needed
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(gstreamer_pipeline(), cv2.CAP_GSTREAMER)
            time.sleep(0.5)
    
        ret, frame_check = self.cap.read()
    
        if not ret:
            self.label.setText("Camera error during identity check.")
            QTimer.singleShot(2000, self.start_standby)
            return
    
        identity_check, _ = detect.detect_verify(frame_check, self.database)
    
        if identity_check == self.identity:
            self.successful_identity = True
            self._start_check_emotion()
            return
    
        self.verification_attempts -= 1
        QTimer.singleShot(500, self._verify_identity)
    def _start_check_emotion(self):
        self.emotion_attempts = 1
        self.detected_emotion = "Unknown"
        self._check_emotion()

    def _check_emotion(self):
        if self.emotion_attempts == 0:
            self._finalize_answer_with_emotion()
            return

        ret, frame = self.cap.read()
        if ret:
            try:
                _, emotion = detect_emotions_in_frame(frame)
                if emotion and len(emotion) > 0:
                    first_emotion = emotion[0]
                    if first_emotion and first_emotion.lower() != "unknown":
                        self.detected_emotion = first_emotion
                        self._finalize_answer_with_emotion()
                        return
            except Exception as e:
                print(f"Emotion detection error: {e}")

        self.emotion_attempts -= 1
        QTimer.singleShot(500, self._check_emotion)


    def get_adult(self):
        adult_name = ""
    
        # Start the camera if it was released
        if self.cap is None or not self.cap.isOpened():
            self.cap = cv2.VideoCapture(gstreamer_pipeline(), cv2.CAP_GSTREAMER)
    
        def play_adult_audio():
            play_audio_file(adult)
    
        while adult_name != "Mr.Ali":
            threading.Thread(target=play_adult_audio).start()
    
            ret, frame = self.cap.read()
            if not ret:
                self.label.setText("Camera error3.")
                self.video_timer.stop()
                return
    
            adult_name, annotated_frame = detect.detect_verify(frame, self.database)
            rgb_frame = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)
            h, w, ch = rgb_frame.shape
            bytes_per_line = ch * w
            qt_img = QImage(rgb_frame.data, w, h, bytes_per_line, QImage.Format_RGB888)
            self.label.setPixmap(QPixmap.fromImage(qt_img).scaled(self.label.width(), self.label.height()))
            QApplication.processEvents()
            time.sleep(0.5)  # slight delay to prevent excessive frame grabbing
    
        self.cap.release()
        self.cap = None
        QTimer.singleShot(5000, self.start_standby)
    def _finalize_answer_with_emotion(self):
        emotion = self.detected_emotion.lower()

        if emotion in ["angry", "fear"]:
            self.label.setText(f"Detected emotion: {emotion}. Getting an adult...")
            self.get_adult()
            QTimer.singleShot(5000, self.start_standby)
            return

        correct = qwen_ar.evaluate_answer(self.answers[self.current_question], self.answer_to_check)
        if correct =='1':
            self.label.setText("اجابه صحيحيه!")
            play_audio_file(correct_path)
            self.score += 1
        else:
            self.label.setText("اجابه خاطئه , برجاء التركيز.")
            play_audio_file(false_ques)

        self.current_question += 1
        QTimer.singleShot(3000, self.ask_question)
if __name__ == '__main__':
    app = QApplication(sys.argv)
    window = AIGUI()
    window.show()
    sys.exit(app.exec_())
```
