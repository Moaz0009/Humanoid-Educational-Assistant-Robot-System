# Image Upload / Face Register

```python
def register_face(name, image_path):
    url = "http://127.0.0.1:8000/upload_image"
    frame=cv2.imread(image_path)
    _, img_encoded = cv2.imencode('.jpg', frame)
    img_bytes = img_encoded.tobytes()
    data = {"name": name}
    response = requests.post(url, files={"image": ("frame.jpg", img_bytes, "image/jpeg")},data=data)
    if response.status_code == 200:
        print("Face detection successful!")
        print("Detected faces:", response.json())
    else:
        print("Error:", response.status_code)
```

```python
register_face('Moaz', 'moaz.jpg')
```

# Face Detection & Verification

```python
import requests
import cv2

def det(frame):
    url = "http://127.0.0.1:8000/detect_verify"
    _, img_encoded = cv2.imencode('.jpg', frame)
    img_bytes = img_encoded.tobytes()
    response = requests.post(url, files={"image": ("frame.jpg", img_bytes, "image/jpeg")})
    if response.status_code == 200:
        print("Face detection successful!")
        print("Detected faces:", response.json())
        return response.json()
    else:
        print("Error:", response.status_code)

# Capture frame from webcam
cap = cv2.VideoCapture(0)  # Use 0 for the default camera

while True:
    ret, frame = cap.read()  # Capture a frame
    if not ret:
        break
    res=det(frame)['faces']
    for (name,cord) in res.items():
        x,y,w,h= cord
        cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)
        cv2.putText(frame, name, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 255), 2)
        cv2.imshow("Frame", frame)  # Show raw frame until processed
    else:
        cv2.imshow("Frame", frame)  # Show raw frame until processed 
    # Send frame for face detection (for example, send every 10th frame)
    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import requests
import cv2
import threading
processed_frame = None
lock = threading.Lock()  # To ensure thread-safe access
def det(frame):
    url = "http://127.0.0.1:8000/detect_verify"
    _, img_encoded = cv2.imencode('.jpg', frame)
    img_bytes = img_encoded.tobytes()
    response = requests.post(url, files={"image": ("frame.jpg", img_bytes, "image/jpeg")})
    if response.status_code == 200:
        print("Face detection successful!")
        print("Detected faces:", response.json())
        return response.json()
    else:
        print("Error:", response.status_code)

def process(fram):
    global processed_frame
    res=det(fram)['faces']
    for (name,cord) in res.items():
        x,y,w,h= cord
        cv2.rectangle(fram, (x, y), (x + w, y + h), (255, 0, 0), 2)
        cv2.putText(fram, name, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 255), 2)
    with lock:
        processed_frame = fram.copy()
counter=0
# Capture frame from webcam
cap = cv2.VideoCapture(0)  # Use 0 for the default camera

while True:
    ret, frame = cap.read()  # Capture a frame
    if ret:
        if counter %2==0:
            try:
                threading.Thread( target=process,args=(frame.copy(),),daemon=True).start()
            except ValueError:
                pass
        counter+=1
    # Show the frame (optional)
    with lock:
        if processed_frame is not None:
            cv2.imshow("Frame", processed_frame)
        else:
            cv2.imshow("Frame", frame)  # Show raw frame until processed
    # Send frame for face detection (for example, send every 10th frame)
    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
