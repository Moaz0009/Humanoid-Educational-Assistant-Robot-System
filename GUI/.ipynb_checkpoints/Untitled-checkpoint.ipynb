```python
import tkinter as tk
from tkinter import Label
from PIL import Image, ImageTk
import cv2
import threading
import time

class RobotGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("Robot Interface")
        self.root.attributes("-fullscreen", True)
        self.root.configure(bg="white")

        self.video_capture = cv2.VideoCapture(0)
        self.camera_running = False

        self.display = Label(self.root, bg="white")
        self.display.pack(expand=True, fill="both")

        # Simulate touchscreen buttons for testing
        self.btn_frame = tk.Frame(self.root, bg="white")
        self.btn_frame.pack(side="bottom", pady=10)

        tk.Button(self.btn_frame, text="Standby", font=("Arial", 20),
                  command=self.show_standby).pack(side="left", padx=20)
        tk.Button(self.btn_frame, text="Camera", font=("Arial", 20),
                  command=self.start_camera).pack(side="left", padx=20)
        tk.Button(self.btn_frame, text="Question", font=("Arial", 20),
                  command=self.show_question).pack(side="left", padx=20)
        tk.Button(self.btn_frame, text="Exit", font=("Arial", 20),
                  command=self.root.destroy).pack(side="left", padx=20)

        self.show_standby()

    def show_standby(self):
        self.stop_camera()
        emoji = "😴"
        self.display.config(text=emoji, font=("Arial", 200), fg="gray", image="", compound="center")

    def start_camera(self):
        self.camera_running = True
        self.display.config(text="", image="", compound="center")
        threading.Thread(target=self.update_camera, daemon=True).start()

    def stop_camera(self):
        self.camera_running = False

    def update_camera(self):
        while self.camera_running:
            ret, frame = self.video_capture.read()
            if not ret:
                break
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            img = Image.fromarray(frame)
            img = img.resize((self.root.winfo_width(), self.root.winfo_height()))
            imgtk = ImageTk.PhotoImage(image=img)
            self.display.imgtk = imgtk
            self.display.config(image=imgtk)
            time.sleep(0.03)  # ~30 FPS

    def show_question(self):
        self.stop_camera()
        self.display.config(image="", text="What is 2 + 2 ?", font=("Arial", 60), fg="black")

        # Show "ANSWER" after 2 seconds
        self.root.after(2000, lambda: self.display.config(text="ANSWER", font=("Arial", 60), fg="blue"))

        # Simulate "EVALUATING" after 4 seconds
        self.root.after(4000, lambda: self.display.config(text="EVALUATING...", font=("Arial", 60), fg="orange"))

        # Simulate result after 6 seconds
        self.root.after(6000, lambda: self.display.config(text="Correct! 🎉", font=("Arial", 60), fg="green"))

if __name__ == "__main__":
    root = tk.Tk()
    app = RobotGUI(root)
    root.mainloop()
```
